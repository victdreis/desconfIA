```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Ler o CSV usando separador e encoding corretos
df = pd.read_csv("dataset_unificado.csv", sep=";", encoding="utf-8-sig")

# 2. Garantir que as colunas existem
if not {'texto', 'label'}.issubset(df.columns):
    raise ValueError(f"Colunas disponíveis: {df.columns.tolist()} — não encontrei 'texto' e 'label'.")

# 3. Criar dataset 1 (30% - texto e label)
dataset1, _ = train_test_split(
    df[['texto', 'label']].dropna(subset=['texto', 'label']),
    test_size=0.70,
    random_state=42
)

# 4. Criar dataset 2 (70% - somente texto)
_, dataset2 = train_test_split(
    df[['texto', 'label']].dropna(subset=['texto', 'label']),
    test_size=0.70,
    random_state=42
)

# 5. Salvar datasets
dataset1.to_csv("dataset_30_texto_label.csv", index=False, sep=";", encoding="utf-8-sig")
dataset2.to_csv("dataset_70_somente_texto.csv", index=False, sep=";", encoding="utf-8-sig")

print("✅ Dataset 1 (30% - texto e label) salvo como dataset_30_texto_label.csv")
print("✅ Dataset 2 (70% - somente texto) salvo como dataset_70_somente_texto.csv")
```
